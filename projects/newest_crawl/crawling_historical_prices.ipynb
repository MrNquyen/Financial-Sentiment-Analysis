# HISTORY

```python
start_date = "2021-12-01"
end_date = "2026-01-22"
history_df = quote.history(start=start_date, end=end_date)
history_df["time"] = pd.to_datetime(history_df["time"])
history_df = history_df[(history_df["time"] >= start_date) & (history_df["time"] <= end_date)]
```
